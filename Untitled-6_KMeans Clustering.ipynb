### KMeans Clustering

K- means is an unsupervised ML Algorithms that groups similar data points into K - Clusters

### The algorithms tries to :
- Minimize distance within a cluster
- Maximize distance between cluster.

### The word:
- k = number of Clusters
- Means = Cluster Center (average value)

### Algorithms:
- Step01: Initialize K Centroid Randomly.
- Step02: Assign Every point to nearest Centroid
- Step03: Update the Centroid position
- Step04: Repeat Step 2 and 3 untill convergence

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
load_iris().target
df = pd.DataFrame(
    load_iris().data,
    columns=load_iris().feature_names
)
df.head()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    df["sepal length (cm)"],
    df["sepal width (cm)"],
    c="teal"
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
k = 3
```

```python
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(df)
```

```python
labels = kmeans.labels_
```

```python
labels
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    df["sepal length (cm)"],
    df["sepal width (cm)"],
    c = labels,
    cmap=  'viridis'
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Predicted Cluster")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    df["sepal length (cm)"],
    df["sepal width (cm)"],
    c = load_iris().target,
    cmap=  'viridis'
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Actual Cluster")
plt.show()
```

```python
centroids = kmeans.cluster_centers_
```

```python
print(centroids)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    df["sepal length (cm)"],
    df["sepal width (cm)"],
    c = labels,
    cmap=  'viridis'
)
plt.scatter(
    centroids[:,0],
    centroids[:,1],
    c = 'red',
    marker='x'
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Predicted Cluster")
plt.show()
```

### WCSS Method(Within-Cluster sum of squares) in KMeans

WCSS is the sum of squared distances between each data points and the centroids of its assigned 

```python
kmeans.inertia_
```

```python
inertia = []
for k in range(1,11):
    model = KMeans(
        n_clusters=k,
        random_state=42
    )
    model.fit(df)
    inertia.append(model.inertia_)
```

```python
inertia
```

```python
plt.figure(figsize=(8,6))
plt.plot(
    range(1,11),
    inertia,
    marker='o'
)
plt.xlabel("Number of cluster(k)")
plt.ylabel("WCSS")
plt.title("Elbow method")
plt.show()
```
